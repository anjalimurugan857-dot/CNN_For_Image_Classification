```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense, Dropout
```

```python
# Image size
IMG_HEIGHT = 128
IMG_WIDTH = 128

# Batch size
BATCH_SIZE = 32
```

```python
train_datagen = ImageDataGenerator(
    rescale=1.0/255.0
)

validation_datagen = ImageDataGenerator(
    rescale=1.0/255.0
)

test_datagen = ImageDataGenerator(
    rescale=1.0/255.0
)
```

```python

train_data = train_datagen.flow_from_directory(
    r"C:\Users\Anjali Murugan\OneDrive\Image_Classification_CNN\train_data",
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode="sparse",
    color_mode="rgb"
)
```

```python
validation_data = validation_datagen.flow_from_directory(
    r"C:\Users\Anjali Murugan\OneDrive\Image_Classification_CNN\validation_data",
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode="sparse",
    color_mode="rgb"
)
```

```python
test_data = test_datagen.flow_from_directory(
   r"C:\Users\Anjali Murugan\OneDrive\Image_Classification_CNN\test_data",
    target_size=(IMG_HEIGHT, IMG_WIDTH),
    batch_size=BATCH_SIZE,
    class_mode="sparse",
    color_mode="rgb",
    shuffle=False
)
```

```python
model = Sequential()

model.add(Conv2D(32,(3, 3),activation="relu",input_shape=(IMG_HEIGHT, IMG_WIDTH, 3)))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Conv2D(64,(3, 3),activation="relu"))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Conv2D(128,(3, 3),activation="relu"))
model.add(MaxPooling2D(pool_size=(2, 2)))

model.add(Flatten())
# Fully Connected Layer
model.add(Dense(256,activation="relu"))
model.add(Dropout(0.3))
# Output Layer
# 5 Classes
model.add(Dense(5,activation="softmax"))
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.summary()
```

```python
history = model.fit(
    train_data,
    validation_data=validation_data,
    epochs=20
)
```

```python
test_loss, test_accuracy = model.evaluate(test_data)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```

```python
model.save("image_classification_cnn.h5")

print("Model Saved Successfully!")
```
